## Configuration

Match `model` to your P1 checkpoint. These defaults describe the reference architecture, not a required student architecture.

Set `base_checkpoint` to the checkpoint selected in P1, such as `expts/<P1 run>/checkpoints/checkpoint-best-metric-model.pth`. Copy its saved `model` settings, then set P2 `model.max_len` to P1 `data.context_length` and `model.num_classes` to your tokenizer's vocabulary size. P2 saves its working configuration to `config_p2.yaml`.

Change the run names for a new experiment. To resume an interrupted P2 run, set its `resume` to true; this restores `last.pth` at an epoch boundary. It does not restart from P1. Keep that run's model, LoRA, data and training settings unchanged. After resumed training, evaluation and adapter export use the current selected checkpoint. Previous outputs are retained in evaluation-history/ and adapter-history/.

`max_examples_per_epoch: null` uses every training pair. A number samples that many pairs without replacement each epoch. `dev_limit: null` evaluates all dev inputs; subset scores must be labeled as such.

````python
%%writefile config_p2.yaml
seed: 11785
data:
  raw_dir: data/webnlg_raw
  prepared_dir: data/webnlg
  download: true
  dev_limit: null
base_checkpoint: expts/YOUR_P1_RUN/checkpoints/checkpoint-best-metric-model.pth
output_dir: expts
runs:
  full: {name: webnlg-full-01, resume: false}
  lora: {name: webnlg-lora-01, resume: false}
model:
  num_layers: 12
  d_model: 512
  num_heads: 8
  num_kv_heads: 2
  d_ff: 2048
  dropout: 0.1
  max_len: 1024
  num_classes: 50257
  weight_tying: true
  use_rope: true
  rope_base: 10000.0
training:
  epochs: 3
  batch_size: 4
  gradient_accumulation_steps: 8
  gradient_clip: 1.0
  max_examples_per_epoch: null
  num_workers: 0
  amp: true
  use_wandb: true # Required for P2 submission. Do not change.
  wandb_project: hw4p2
  wandb_entity: null
optimizer:
  lr: 0.0001
  betas: [0.9, 0.95]
  weight_decay: 0.01
scheduler:
  name: constant
  cosine: {T_max: 3, eta_min: 0.00001}
lora:
  rank: 8
  alpha: 16
  dropout: 0.05
  targets: [q_proj, v_proj]
generation:
  max_new_tokens: 128
submission:
  # Choose the model used for inference and Kaggle submission:
  # "full" loads the Full SFT checkpoint; "lora" loads the LoRA adapter on top of the P1 checkpoint.
  model: lora
  # Competition test inputs downloaded by the Kaggle data cell.
  # Change only if you stored test.csv elsewhere.
  test_file: data/test.csv
  predictions_file: results.csv
````

````python
config = yaml.safe_load(Path("config_p2.yaml").read_text())
device = "cuda" if torch.cuda.is_available() else "cpu"

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed(config["seed"])
for key in ("raw_dir", "prepared_dir"):
    config["data"][key] = os.path.expandvars(os.path.expanduser(config["data"][key]))
config["base_checkpoint"] = os.path.expandvars(os.path.expanduser(config["base_checkpoint"]))
config["output_dir"] = os.path.expandvars(os.path.expanduser(config["output_dir"]))
if os.environ.get("SLURM_JOB_ID") and config["output_dir"] == "expts":
    import getpass
    config["output_dir"] = f"/u/{getpass.getuser()}/hw4p2_runs"
print("Checkpoint:", config["base_checkpoint"])
print("Outputs:", config["output_dir"])
if config["training"]["use_wandb"]:
    import wandb
    # Uses WANDB_API_KEY or an interactive hidden prompt. Do not paste a key here.
    wandb.login()
````

## Data

### Prepare the WebNLG Training and Validation Data

Use the English v3.0 WebNLG `train.tar.gz` and `dev.tar.gz` archives in `raw_dir`, or set `download: true`. Alternatively, point `prepared_dir` to a local/Drive copy of the prepared data.

Preparation groups references by triples and removes training graphs found in dev. The public WebNLG dev split is for development, not a secret test set. WebNLG v3.0 uses [CC BY-NC-SA 4.0](https://gitlab.com/shimorina/webnlg-dataset/-/blob/master/release_v3.0/README.md#license). Please credit Gardent, Shimorina, Narayan and Perez-Beltrachini, [Creating Training Corpora for NLG Micro-Planners](https://aclanthology.org/P17-1017/), ACL 2017.

````python
from pathlib import Path
import json
import random
from hw4lib.utils.webnlg import prepare_webnlg, load_webnlg_records
````

````python
data_dir = Path(config["data"]["prepared_dir"])
if not data_dir.exists():
    info = prepare_webnlg(Path(config["data"]["raw_dir"]), data_dir, download=config["data"]["download"])
else:
    info = json.loads((data_dir / "data_info.json").read_text())
train_records = load_webnlg_records(data_dir / "train.jsonl")
dev_records = load_webnlg_records(data_dir / "dev.jsonl")
all_dev_count = len(dev_records)
limit = config["data"]["dev_limit"]
if limit is not None:
    indices = sorted(random.Random(config["seed"]).sample(range(all_dev_count), min(limit, all_dev_count)))
    dev_records = [dev_records[i] for i in indices]
print(info)
print(f"Evaluating {len(dev_records)} / {all_dev_count} dev inputs")
print(json.dumps(train_records[0], indent=2, ensure_ascii=False))
````

````python
tokenizer = GPT2Tokenizer()
# Over-context examples raise an error; facts and reference text are not truncated.
train_data = SFTDataset(train_records, tokenizer, config["model"]["max_len"], expand_references=True)
dev_data = SFTDataset(dev_records, tokenizer, config["model"]["max_len"], expand_references=False)
sampler = EpochWindowSampler(train_data, max_windows=config["training"]["max_examples_per_epoch"], seed=config["seed"])
train_loader = DataLoader(train_data, batch_size=config["training"]["batch_size"], sampler=sampler,
                          collate_fn=train_data.collate_fn, num_workers=config["training"]["num_workers"])
dev_loader = DataLoader(dev_data, batch_size=config["training"]["batch_size"], shuffle=False,
                        collate_fn=dev_data.collate_fn, num_workers=config["training"]["num_workers"])
print("Training pairs:", len(train_data), "| pairs per epoch:", len(sampler))
print("Dev loss: first reference per input. Generation: all references used for scoring.")
batch = next(iter(dev_loader))
print("Inputs:", tuple(batch["input_ids"].shape), "| supervised targets:", int((batch["labels"] != -100).sum()))
````

## Experiment setup

Load the P1 weights and create a fresh optimizer for each run. A cosine schedule, if selected, advances by optimizer updates using the loader's length. Set `T_max` to the intended number of epochs.

W&B logging is required for P2 submission. Keep `training.use_wandb: true`.

````python
from hw4lib.utils.sft_experiment import save_wandb_run

def make_trainer(mode):
    set_seed(config["seed"])
    model = DecoderOnlyTransformer(**config["model"])
    load_base_weights(model, config["base_checkpoint"])
    if mode == "lora":
        print("LoRA projections:", inject_lora(model, **config["lora"]))
    run_config = copy.deepcopy(config)
    run_config["training"]["resume"] = config["runs"][mode]["resume"]
    run_config["experiment"] = mode
    run_name = config["runs"][mode]["name"]
    
    run_config["training"]["use_wandb"] = True
    if run_config["training"]["resume"]:
        run_dir = (
            Path(run_config["output_dir"]).expanduser().resolve()
            / run_name
        )
        run_file = run_dir / "wandb_run.json"

        if not run_file.is_file():
            raise FileNotFoundError(
                f"Cannot resume W&B: {run_file} is missing."
            )

        saved = json.loads(run_file.read_text())
        entity, project, run_id = saved["path"].split("/")

        run_config["training"]["wandb_entity"] = entity
        run_config["training"]["wandb_project"] = project
        run_config["training"]["wandb_run_id"] = run_id
    else:
        run_config["training"]["wandb_run_id"] = None
        
    trainer = SFTTrainer(model, tokenizer, run_config, run_name, str(PROJECT_ROOT / "config_p2.yaml"), device=device)
    parameters = [p for p in trainer.model.parameters() if p.requires_grad]
    trainer.optimizer = torch.optim.AdamW(parameters, **config["optimizer"])
    if config["scheduler"]["name"] == "cosine":
        trainer.scheduler = create_scheduler(trainer.optimizer, config["scheduler"], train_loader,
            gradient_accumulation_steps=config["training"]["gradient_accumulation_steps"])
    elif config["scheduler"]["name"] != "constant":
        raise ValueError("Choose constant or cosine for this notebook")
    if run_config["training"]["resume"]:
        trainer.load_checkpoint("last.pth")
    counts = {"total": sum(p.numel() for p in trainer.model.parameters()),
              "trainable": sum(p.numel() for p in parameters)}
    print(mode, counts)
    counts_file = trainer.expt_root / "parameter_counts.json"
    if not counts_file.exists():
        counts_file.write_text(json.dumps(counts, indent=2))
    if trainer.use_wandb:
        save_wandb_run(trainer)
    return trainer

# DO NOT MODIFY: score predictions with the same corpus metric in every run.
def record_evaluation(trainer, prefix=""):
    metrics, predictions = evaluate_and_export(trainer, dev_loader, dev_records,
        available_inputs=all_dev_count, seed=config["seed"], prefix=prefix)
    print(json.dumps({k: v for k, v in metrics.items() if k != "evaluation"}, indent=2))
    return metrics, predictions
````

## Base evaluation and full fine-tuning

The **base** is your P1 model before SFT. **Full fine-tuning** updates all its parameters. Select the checkpoint by dev Kaggle score: (BLEU + ROUGE-L) / 2.

````python
full_trainer = make_trainer("full")
full_dir = full_trainer.expt_root
if not config["runs"]["full"]["resume"]:
    base_metrics, base_predictions = record_evaluation(full_trainer, prefix="base_")
else:
    base_metrics = json.loads((full_dir / "base_metrics.json").read_text())
full_trainer.train(train_loader, dev_loader, dev_records)
````

````python
load_base_weights(full_trainer.model, full_trainer.best_model_path)
full_metrics, full_predictions = record_evaluation(full_trainer)
full_trainer.cleanup()
del full_trainer
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
````

## Results

Kaggle score = (BLEU + ROUGE-L) / 2, with both components on a 0–100 scale. Use this score for checkpoint selection and the Kaggle leaderboard. ROUGE-L uses the best reference F1 for each input, averaged over inputs, with no stemming. Response CE is negative log likelihood per response token; response PPL is `exp(CE)`.

Check that response loss falls, inspect omissions or invented facts, and compare full fine-tuning and LoRA to the base. These dev scores select checkpoints; they are not independent final-test estimates. No passing score has been set here.

````python
### Compare Saved Evaluation Results (Base, Full SFT, and LoRA metrics)
from pathlib import Path
import pandas as pd

output_root = Path(config["output_dir"]).expanduser().resolve()
full_dir = output_root / config["runs"]["full"]["name"]
lora_dir = output_root / config["runs"]["lora"]["name"]

metric_files = [
    ("P1 base", full_dir / "base_metrics.json"),
    ("Full SFT", full_dir / "metrics.json"),
    ("LoRA", lora_dir / "metrics.json"),
]
metric_keys = (
    "kaggle_score",
    "bleu",
    "rouge_l",
    "response_ce",
    "response_ppl",
)

rows = []
for name, path in metric_files:
    if not path.is_file():
        print(f"Note: {name} results are not available yet.")
        continue
    import json
    metrics = json.loads(path.read_text())
    rows.append({
        "model": name,
        **{key: metrics.get(key) for key in metric_keys},
    })
if rows:
    display(pd.DataFrame(rows))
else:
    print("No saved evaluation results found. Run an evaluation first.")
````

````python
# Inspect Saved Dev Predictions
import csv
from pathlib import Path

root = Path(config["output_dir"]).expanduser()
saved = {}
for name, mode in [("Full SFT", "full"), ("LoRA", "lora")]:
    path = root / config["runs"][mode]["name"] / "dev_predictions.csv"
    with path.open(encoding="utf-8", newline="") if path.is_file() else __import__("contextlib").nullcontext(None) as f:
        saved[name] = {r["id"]: r["prediction"] for r in csv.DictReader(f)} if f else {}

for row in dev_records[:3]:
    print("Triples:", row["triples"])
    print("References:", row["references"])
    for name, predictions in saved.items():
        print(f"{name}:", predictions.get(row["id"], "No saved prediction for this example."))
    print()
````

## Inference

````python
# Clear the previous inference and packaging state.
predictor = None
MODEL = None
results_df = None
KAGGLE_SUBMISSION_MESSAGE = None
code_archive = None

from pathlib import Path
import os
import csv
import json
import yaml
from hw4lib.data import GPT2Tokenizer
from hw4lib.trainers.sft_trainer import SFTTrainer
from hw4lib.utils.sft_experiment import load_selected_model

PROJECT_ROOT = Path.cwd().resolve()
CONFIG_PATH = PROJECT_ROOT / "config_p2.yaml"

# Use the saved configuration in a new session.
# Keep the current configuration when continuing in the same session.
if "config" not in globals():
    config = yaml.safe_load(CONFIG_PATH.read_text())

def resolve_project_path(value):
    path = Path(
        os.path.expandvars(os.path.expanduser(str(value)))
    )
    if not path.is_absolute():
        path = PROJECT_ROOT / path
    return path.resolve()

config["base_checkpoint"] = str(
    resolve_project_path(config["base_checkpoint"])
)
config["output_dir"] = str(
    resolve_project_path(config["output_dir"])
)

if not config["submission"].get("test_file"):
    raise ValueError(
        "Set submission.test_file to the competition test CSV."
    )

test_path = resolve_project_path(
    config["submission"]["test_file"]
)
full_dir = (
    Path(config["output_dir"]) / config["runs"]["full"]["name"]
)
lora_dir = (
    Path(config["output_dir"]) / config["runs"]["lora"]["name"]
)

device = "cuda" if torch.cuda.is_available() else "cpu"

inference_model = load_selected_model(
    config, full_dir, lora_dir
).to(device)

# Inference only: do not initialize a new training run.
predictor = object.__new__(SFTTrainer)
predictor.model = inference_model
predictor.tokenizer = GPT2Tokenizer()
predictor.config = config
predictor.device = device

with test_path.open(encoding="utf-8", newline="") as file:
    test_records = list(csv.DictReader(file))

for row in test_records:
    row["triples"] = json.loads(row["triples"])

print("Selected model:", config["submission"]["model"])
print("Test inputs:", len(test_records))
````
